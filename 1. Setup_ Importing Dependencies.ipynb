Purpose: The purpose of this notebook is to check for library dependencies and import the appropriate versions\. 

Inputs
1\. requirements\.txt: text file listing all library dependencies and their version numbers\. 

Outputs
None

```
%%bash
# Library dependencies and their version numbers are listed in requirements.txt.
if test -f requirements.txt
  then
    pip install -qq -r ./requirements.txt
    echo 'Dependencies installed.'
  else echo "There's no requirements.txt, so nothing to install."
fi
```

```
import sys
import platform

print("Python:", sys.version)
print("Platform:", platform.platform())
print("Executable:", sys.executable)
```
